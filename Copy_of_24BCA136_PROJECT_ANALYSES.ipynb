```python
df.columns = df.columns.str.strip()
print("Available columns after stripping whitespace:")
print(df.columns.tolist())

# Example: Calculate the mean of an existing numerical column, e.g., 'src_bytes'
# First, ensure 'src_bytes' is numeric, coercing errors to NaN
df['src_bytes'] = pd.to_numeric(df['src_bytes'], errors='coerce')
# Then, calculate the mean, ignoring NaN values
avg_src_bytes = df['src_bytes'].mean()
print(f"\nMean of 'src_bytes': {avg_src_bytes}")
```

```python
df['src_bytes'] = df['src_bytes'].fillna(avg_src_bytes)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.bar(df['label'],df['src_bytes'],color="purple")
plt.xlabel("label")
plt.ylabel("src_bytes")
plt.xticks(rotation=50)
plt.title("Label vs src_bytes")
plt.show()
```

```python
df['dst_bytes'] = pd.to_numeric(df['dst_bytes'], errors='coerce')
df['src_bytes'] = pd.to_numeric(df['src_bytes'], errors='coerce')

plt.scatter(df['dst_bytes'], df['src_bytes'], color="purple")
plt.xlabel("Destination Bytes")
plt.ylabel("Source Bytes")
plt.xticks(rotation=50)
plt.title("Source Bytes vs Destination Bytes")
plt.show()
```

```python
df['src_bytes'] = pd.to_numeric(df['src_bytes'], errors='coerce')
plt.hist(df['src_bytes'])
plt.xlabel("Source Bytes")
plt.ylabel("Frequency")
plt.xticks(rotation=50)
plt.title("Distribution of Source Bytes")
plt.show()
```

```python
label_counts = df['label'].value_counts()
plt.pie(label_counts, labels=label_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Labels')
plt.show()
```
